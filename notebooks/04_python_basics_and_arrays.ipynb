**Intentional error.** `* *` with a space is not the same as `**`. Python reads it as two
multiplication operators in a row and raises a `SyntaxError`.

```python
# Intentional SyntaxError - the space splits ** into two * operators
3 * * 4
```

```python
# Parentheses override precedence
1 + 2 * (3 * 4 * 5 / 6) ** 3 + 7 + 8 - 9 + 10
```

## Numbers: `int` and `float`

```python
# int values have no fractional part and can be arbitrarily large
print(2)
print(1 + 3)
print(-1234567890000000000)
```

```python
# float values have a decimal point; / always returns a float, even 3 / 1
print(1.2)
print(1.5 + 2)
print(3 / 1)
print(-12345678900000000000.0)
```

**Intentional edge cases.** Floats have limited range and precision. Going past the
largest float gives `inf` (overflow) and going below the smallest gives `0.0` (underflow).

```python
print(2e306 * 10)    # still within range
print(2e306 * 100)   # overflow -> inf
print(2e-322 / 10)   # tiny "subnormal" number, precision is already lost
print(2e-322 / 100)  # underflow -> 0.0
```

```python
# Floats keep only about 15-16 significant digits, so the extra digits are dropped
0.6666666666666666 - 0.6666666666666666123456789
```

```python
# Rounding error: sqrt(2) * sqrt(2) is not exactly 2
print(2 ** 0.5)
print((2 ** 0.5) * (2 ** 0.5))
print((2 ** 0.5) * (2 ** 0.5) - 2)
```

## Growth rates

```python
# Growth rate = (changed - initial) / initial, which is the same as changed / initial - 1
initial = 2766000
changed = 2814000
print((changed - initial) / initial)
print((changed / initial) - 1)
```

```python
# The rate does not depend on the units: dividing both values by 1000 gives the same answer
initial = 2766
changed = 2814
(changed / initial) - 1
```

```python
initial = 2.37
changed = 3.38
(changed / initial) - 1
```

```python
initial = 287.6
changed = 314.1
(changed / initial) - 1
```

```python
# A negative growth rate means the value went down
initial = 15.3
changed = 11.9
(changed / initial) - 1
```

```python
# Compounding 3.5% for ten years, written out the long way...
1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 * 1.035 - 1
```

```python
# ...and with an exponent
annual_growth_rate = 0.035
ten_year_growth_rate = (1 + annual_growth_rate) ** 10 - 1
ten_year_growth_rate
```

```python
# Reverse it: the annual rate that turns `initial` into `changed` over t years
initial = 2.37
changed = 3.38
t = 10
(changed / initial) ** (1 / t) - 1
```

## Arrays

The *Inferential Thinking* textbook builds arrays with `make_array` from the `datascience`
package, which is a thin wrapper around `np.array`. Plain NumPy is used here so the notebook
has no extra dependency.

```python
import numpy as np

# Global temperature highs (°C) relative to a baseline
baseline_high = 14.48
highs = np.array([baseline_high - 0.880,
                  baseline_high - 0.093,
                  baseline_high + 0.105,
                  baseline_high + 0.684])
highs
```

```python
# Average using built-in functions
sum(highs) / len(highs)
```

```python
# Arrays can hold strings too
english_parts_of_speech = np.array(["noun", "pronoun", "verb", "adverb",
                                    "adjective", "conjunction", "preposition", "interjection"])
english_parts_of_speech
```

```python
# Arithmetic applies to every element: convert all highs from °C to °F at once
(9 / 5) * highs + 32
```

```python
# Array methods
print(highs.size)
print(highs.sum())
print(highs.mean())
```

```python
# Differences between consecutive elements
np.diff(highs)
```

### Approximating π with the Leibniz series

π = 4 × (1 − 1/3 + 1/5 − 1/7 + ...). The positive terms have denominators 1, 5, 9, ...
and the negative terms have denominators 3, 7, 11, ...

```python
positive_term_denominators = np.arange(1, 10000, 4)
positive_terms = 1 / positive_term_denominators
negative_terms = 1 / (positive_term_denominators + 2)

4 * (sum(positive_terms) - sum(negative_terms))
```

## Element-wise operations on two arrays

```python
baseline_low = 3.00
lows = np.array([baseline_low - 0.872, baseline_low - 0.629,
                 baseline_low - 0.126, baseline_low + 0.728])
lows
```

```python
# The long way: subtract element by element with .item(i)
np.array([
    highs.item(0) - lows.item(0),
    highs.item(1) - lows.item(1),
    highs.item(2) - lows.item(2),
    highs.item(3) - lows.item(3),
])
```

```python
# The array way: arrays of the same length are subtracted element-wise
highs - lows
```

### Approximating π with the Wallis product

π = 2 × (2/1 × 2/3) × (4/3 × 4/5) × (6/5 × 6/7) × ...

```python
even = np.arange(2, 1000001, 2)
one_below_even = even - 1
one_above_even = even + 1

2 * np.prod(even / one_below_even) * np.prod(even / one_above_even)
```
